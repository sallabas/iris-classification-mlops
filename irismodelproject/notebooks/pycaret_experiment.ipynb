```python
from pycaret.datasets import get_data
from pycaret.classification import *

data = get_data("iris")
data.head()
```

```python
clf = setup(
    data=data,
    target="species",
    session_id=42,
    fold=5,
    normalize=True,
    verbose=True
)
```

```python
best_model = compare_models()
```

```python
best_model
evaluate_model(best_model)
```

```python
final_model = finalize_model(best_model)
```

```python
save_model(final_model, "pycaret_best_model")
```

```python
loaded_model = load_model("pycaret_best_model")
loaded_model
```

```python
sample = data.drop("species", axis=1).iloc[[0]]
predict_model(loaded_model, data=sample)
```
